```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("portfolio.html", "r", encoding = "utf-8")as file1:
  html = file1.read()
  soup = BeautifulSoup(html, "html.parser")
  print(soup.prettify())
```

```python
nav_items = soup.select("nav .nav-container, li, a")
nav_data = [{"Section":item.text.strip(),"Link":item.get("href")}for item in nav_items]
nav_data
```

```python
for i in nav_items:
  dict_items = {"section" : i.text.strip(), "Link": i.get("href")}
  print(dict_items)
```
